# Vector Databases for RAG

> **From an in-memory prototype to a real searchable vector store with Qdrant Cloud.**

We generated embeddings for our documents, embedded a user's query, calculated similarity scores, and returned the top results.

```text
User Query
    ↓
Query Embedding
    ↓
Similarity Search
    ↓
Ranked Results
```

```text
Documents
    ↓
Embeddings
    ↓
Qdrant Cloud
    ↓
Vector Search
    ↓
Retrieved Chunks
```

The goal is not simply to learn how to call the Qdrant API.

The goal is to understand **what a vector database contributes to a RAG system and why it exists**.

# 1. Why Do We Need a Vector Database?

In the previous notebook, our retrieval code was essentially:

```python
scores = document_embeddings @ query_embedding
```

We then sorted those scores and selected the top results.

That is perfectly fine for a small experiment.

But imagine:

```text
10 chunks
100 chunks
1,000 chunks
10,000 chunks
1,000,000 chunks
```

Keeping everything in Python memory and calculating every comparison ourselves becomes increasingly impractical.

We therefore want infrastructure designed for:

- Storing vectors
- Searching vectors efficiently
- Storing associated metadata
- Filtering records
- Updating records
- Deleting records
- Managing larger collections

That's where a vector database comes in.

# 2. What Is a Vector Database?

A vector database is a system designed to store and retrieve vector representations efficiently.

In our RAG system, a stored record conceptually looks like:

```text
┌──────────────────────────────────────┐
│               POINT                  │
├──────────────────────────────────────┤
│ ID                                   │
│ Vector                               │
│ Payload / Metadata                   │
└──────────────────────────────────────┘
```

For example:

```python
{
    "id": "refund-policy-001",
    "vector": [0.021, -0.184, 0.731, ...],
    "payload": {
        "text": "Customers can request...",
        "source": "refund_policy.pdf",
        "page": 4
    }
}
```

The vector is useful for similarity search.

The payload gives us the information we actually want to retrieve and use.

This distinction will be important throughout the course.

# 6. Collections

A **collection** is a named collection of points that share a vector configuration.

Conceptually:

```text
Qdrant Cloud
     │
     ├── collection: customer-support
     │
     ├── collection: product-docs
     │
     └── collection: research-docs
```

For this tutorial, we'll create:

```text
rag-tutorial
```

The collection will contain the embeddings and metadata for our knowledge base.

# 7. Vector Dimensions

Before creating the collection, we need to know the dimensionality of our embedding vectors.

We can inspect our embedding matrix:

```python
document_embeddings.shape
```

If the output is:

```text
(number_of_chunks, embedding_dimension)
```

then the second value is the vector size.

For example:

```python
vector_size = document_embeddings.shape[1]

print(vector_size)
```

The collection must be configured with the same vector dimension as the embedding model produces.

This gives us an important rule:

> **The vector size configured in the database must match the dimension of the embeddings we upload.**

# 8. Distance Metrics

A vector database needs to know how vectors should be compared.

Common choices include:

- Cosine similarity
- Dot product
- Euclidean distance

Our previous notebook normalized the embeddings and used a dot product:

```python
document_embeddings @ query_embedding
```

We'll use **cosine distance** in Qdrant for this tutorial.

The important thing is consistency.

The similarity behavior used during retrieval depends on:

```text
Embedding Model
       +
Distance Metric
```

Choosing a metric is therefore part of the retrieval design.

# 9. Create the Collection

Import the Qdrant models we need:

```python
from qdrant_client.models import Distance, VectorParams
```

Create the collection:

```python
collection_name = "rag-tutorial"

client.create_collection(
    collection_name=collection_name,
    vectors_config=VectorParams(
        size=vector_size,
        distance=Distance.COSINE,
    ),
)
```

We have now created an empty vector collection.

Conceptually:

```text
Qdrant Collection
       │
       ├── Vector configuration
       │       ├── Dimension
       │       └── Distance
       │
       └── Points
```

The collection is ready to receive our embeddings.

# 10. What Is a Point?

In Qdrant, the data we store is organized into **points**.

A point combines an identifier, vector data, and optional payload.

For our RAG application:

```text
Point
│
├── ID
│
├── Vector
│
└── Payload
      ├── text
      ├── document_id
      ├── source
      ├── page
      └── other metadata
```

This maps naturally to the records we created in the indexing notebook.

# 11. Prepare Our Documents

Let's recreate our small knowledge base.

```python
documents = [
    {
        "id": "refund_policy",
        "text": "Customers can request a refund within 30 days of purchase."
    },
    {
        "id": "refund_processing",
        "text": "Approved refunds are normally processed within 7 business days."
    },
    {
        "id": "refund_condition",
        "text": "Products must be returned in their original condition to qualify for a refund."
    },
    {
        "id": "shipping_standard",
        "text": "Standard shipping normally takes between 3 and 5 business days."
    },
    {
        "id": "shipping_express",
        "text": "Express shipping normally takes between 1 and 2 business days."
    },
    {
        "id": "shipping_tracking",
        "text": "Customers can track orders using the tracking number provided by email."
    },
    {
        "id": "support_hours",
        "text": "Customer support is available Monday through Friday from 9 AM to 5 PM."
    }
]
```

Generate embeddings:

```python
texts = [document["text"] for document in documents]

document_embeddings = embedding_model.encode(
    texts,
    normalize_embeddings=True
)
```

Now we have everything required to create our points.

# 12. Build Qdrant Points

We'll use Qdrant's `PointStruct`.

```python
from qdrant_client.models import PointStruct
```

Create the points:

```python
points = []

for document, embedding in zip(documents, document_embeddings):
    points.append(
        PointStruct(
            id=document["id"],
            vector=embedding.tolist(),
            payload={
                "text": document["text"],
                "document_id": document["id"],
            },
        )
    )
```

Notice what we're storing.

The vector is stored as the searchable representation.

The payload stores the information we want to retrieve alongside it.

# 13. Upload the Points

Now send the points to Qdrant Cloud:

```python
client.upsert(
    collection_name=collection_name,
    points=points,
)
```

`upsert` means that the points will be inserted or updated based on their IDs.

Our architecture is now:

```text
Documents
    ↓
Embeddings
    ↓
PointStruct
    ↓
Qdrant Cloud
```

Let's verify the collection:

```python
client.get_collection(collection_name)
```

Our vector data is no longer only inside the notebook's Python process.

It lives in our Qdrant Cloud collection.

# 14. Search the Vector Database

Now comes the important part.

We want to answer:

> **How long do I have to request a refund?**

Create the query embedding:

```python
query = "How long do I have to request a refund?"

query_embedding = embedding_model.encode(
    query,
    normalize_embeddings=True
)
```

We can now search the collection.

```python
search_results = client.query_points(
    collection_name=collection_name,
    query=query_embedding.tolist(),
    limit=3,
).points
```

Inspect the results:

```python
for result in search_results:
    print(f"Score: {result.score:.4f}")
    print(f"ID: {result.id}")
    print(result.payload["text"])
    print("---")
```

We have replaced our in-memory similarity search with a query against Qdrant Cloud.

# 15. What Just Happened?

Previously:

```text
Query
  ↓
Embedding
  ↓
NumPy
  ↓
Compare every vector
```

Now:

```text
Query
  ↓
Embedding
  ↓
Qdrant Cloud
  ↓
Vector Search
  ↓
Top-K Points
```

The conceptual retrieval operation hasn't changed.

We still have:

```text
Query Embedding
       ↓
Similarity Search
       ↓
Ranked Results
```

What changed is the **storage and search infrastructure**.

This distinction is important:

> **A vector database doesn't replace retrieval. It provides infrastructure that makes vector retrieval practical at scale.**

# 16. Payloads Are More Than Text

So far our payload contains:

```python
{
    "text": "...",
    "document_id": "..."
}
```

In a real application, we may store much more:

```python
{
    "text": "...",
    "document_id": "refund-policy",
    "source": "refund_policy.pdf",
    "page": 4,
    "section": "Refund Eligibility",
    "version": 3,
    "department": "customer-service"
}
```

This information can support:

- Citations
- Filtering
- Access control
- Debugging
- Document versioning
- Provenance

The vector tells us **where the information is in the embedding space**.

The payload tells us **what the information is and where it came from**.

# 17. Metadata Filtering

Vector similarity isn't the only way we may want to search.

Suppose our knowledge base contains documents from multiple departments.

We might want:

> "Search only customer-service documents."

This is where metadata filtering becomes useful.

Conceptually:

```text
Query
  ↓
Vector Search
  +
Metadata Filter
  ↓
Relevant Results
```

For example, if our payload contains:

```python
"department": "customer-service"
```

we can filter results based on that field.

We'll build more advanced filtering examples as our metadata schema becomes richer.

# 18. Why Filtering Matters in RAG

Filtering isn't just a performance feature.

It can affect correctness.

Imagine a company has:

```text
Public policy
Internal policy
Archived policy
Customer-specific policy
```

A user shouldn't necessarily retrieve all of them.

Without appropriate filtering, the system could retrieve information that:

- Belongs to another tenant
- Is outdated
- Is outside the user's permissions
- Applies to a different product
- Comes from the wrong region

This is why metadata and retrieval design eventually become part of the application's security model.

We'll explore these concerns more deeply later in the course.

# 19. Updating a Point

Documents change.

Suppose the refund policy changes from:

```text
30 days
```

to:

```text
60 days
```

Because we used stable point IDs, we can update the existing point.

First create the new embedding:

```python
updated_text = (
    "Customers can request a refund within 60 days of purchase."
)

updated_embedding = embedding_model.encode(
    updated_text,
    normalize_embeddings=True
)
```

Then upsert the updated point:

```python
client.upsert(
    collection_name=collection_name,
    points=[
        PointStruct(
            id="refund_policy",
            vector=updated_embedding.tolist(),
            payload={
                "text": updated_text,
                "document_id": "refund_policy",
            },
        )
    ],
)
```

The old representation has now been replaced for that point ID.

This illustrates an important indexing principle:

> **Changing the source content generally means changing its embedding too.**

# 20. Deleting a Point

If a document should no longer be searchable, we need to remove its indexed representation.

For example:

```python
client.delete(
    collection_name=collection_name,
    points_selector=["refund_policy"],
)
```

After deletion, that point should no longer participate in vector retrieval.

Document lifecycle therefore becomes:

```text
Create
  ↓
Index
  ↓
Update
  ↓
Re-index
  ↓
Delete / Archive
```

A production RAG system needs to handle this lifecycle reliably.

# 21. Connecting Qdrant to Our RAG Pipeline

We can now connect everything we've learned.

The indexing side:

```text
Documents
    ↓
Chunking
    ↓
Embeddings
    ↓
Qdrant Cloud
```

The query side:

```text
User Query
    ↓
Query Embedding
    ↓
Qdrant Search
    ↓
Retrieved Points
    ↓
Payload Text
    ↓
Context
    ↓
LLM
```

The complete system becomes:

```text
                    INDEXING
                        │
                        ▼
                    Documents
                        │
                        ▼
                    Chunking
                        │
                        ▼
                    Embeddings
                        │
                        ▼
                  ┌───────────┐
                  │  Qdrant   │
                  │   Cloud   │
                  └─────┬─────┘
                        │
                        │
                     QUERY
                        │
                        ▼
                    User Query
                        │
                        ▼
                 Query Embedding
                        │
                        ▼
                  Vector Search
                        │
                        ▼
                 Retrieved Points
                        │
                        ▼
                     Context
                        │
                        ▼
                       LLM
                        │
                        ▼
                     Answer
```

This is now a real RAG retrieval architecture rather than an in-memory demonstration.

# 22. Vector Database vs RAG

It's important not to confuse these concepts.

A vector database is **not** a RAG system.

You can have:

```text
Vector Database
```

without an LLM.

And you can build some retrieval systems without a dedicated vector database.

RAG is the larger architecture:

```text
RAG
│
├── Document ingestion
├── Chunking
├── Embeddings
├── Indexing
├── Retrieval
├── Context construction
└── Generation
```

Qdrant is one component inside that architecture.

```text
RAG System
     │
     └── Vector Database
             │
             └── Qdrant
```

Keeping these concepts separate will prevent a lot of confusion later.

# 23. What a Vector Database Solves

Our original prototype had to manage:

```text
Embeddings in Python
Similarity calculations
Metadata
Storage
Updates
Deletes
```

Qdrant gives us infrastructure for these storage and retrieval concerns.

The larger RAG system still has to solve:

```text
Document ingestion
Chunking
Embedding selection
Query processing
Retrieval strategy
Reranking
Context construction
Generation
Evaluation
```

So introducing Qdrant doesn't magically make our RAG system production-ready.

It gives us a proper foundation for one important part of it: **vector storage and search**.

# Key Takeaways

1. A vector database stores and searches vector representations efficiently.
2. Our in-memory NumPy retriever is useful for learning but does not provide scalable storage infrastructure.
3. Qdrant Cloud lets us use a real vector database from both local JupyterLab and Google Colab.
4. A Qdrant point combines an ID, vector, and optional payload.
5. The vector is used for similarity search.
6. Payloads preserve text and metadata associated with the vector.
7. The collection's vector dimension must match the embedding dimension.
8. The distance metric determines how vectors are compared.
9. Metadata filtering can constrain retrieval to the appropriate subset of knowledge.
10. Updating source content generally requires updating its embedding.
11. A vector database is **one component of a RAG system**, not the RAG system itself.

The mental model to keep is:

```text
                  RAG
                   │
        ┌──────────┴──────────┐
        ↓                     ↓
    Indexing                Querying
        │                     │
        ↓                     ↓
   Embeddings              Retrieval
        │                     │
        ↓                     ↓
   Qdrant Cloud          Retrieved Points
                              │
                              ↓
                           Context
                              │
                              ↓
                             LLM
                              │
                              ↓
                           Answer
```

# What's Next?

We now have a proper vector store.

But our retrieval system still has a significant limitation:

```text
Query
  ↓
Dense Embedding
  ↓
Vector Search
```

What happens when the user searches for an exact product code?

Or a policy number?

Or a person's name?

Or a technical identifier?

Pure semantic retrieval isn't always enough.

The next step is to combine different retrieval signals:

```text
                 Query
                   │
          ┌────────┴────────┐
          ↓                 ↓
    Dense Retrieval    Keyword Retrieval
          ↓                 ↓
       Candidates        Candidates
          └────────┬────────┘
                   ↓
            Combined Ranking
```
