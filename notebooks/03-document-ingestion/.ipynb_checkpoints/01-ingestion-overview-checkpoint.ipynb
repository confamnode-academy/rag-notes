## 1. The Two Flows of RAG

The RAG stack can be understood as two major flows:

1. **Ingestion flow** — prepares external knowledge for future queries.
2. **Query flow** — processes a user's question and uses the prepared knowledge to generate a response.

Conceptually:

```text
                 RAG SYSTEM
                     │
          ┌──────────┴──────────┐
          ↓                     ↓
    INGESTION FLOW         QUERY FLOW
          │                     │
   Prepare knowledge       Find knowledge
          │                     │
          └──────────┬──────────┘
                     ↓
                  Answer
```

The ingestion flow is normally triggered when new data becomes available or existing data needs to be updated. The query flow is triggered whenever a user sends a query.

## 2. Where Document Ingestion Fits

The source material describes the core ingestion flow as:

```text
Parsing → Chunking → Embedding → Indexing
```

For this course, we deliberately separate those stages into different modules so that each engineering problem can be studied properly.

Therefore, **Module 03 focuses on the work that happens before chunking**:

```text
Raw Document
    ↓
Validate & Register
    ↓
Parse
    ↓
OCR if Required
    ↓
Clean & Normalize
    ↓
Preserve Structure
    ↓
Attach Metadata & Provenance
    ↓
Validate Output
    ↓
Ready for Chunking
```

Chunking itself belongs to **Module 04**.

## 3. What Can We Ingest?

RAG systems can receive knowledge from several types of sources, including:

- Files stored locally or in cloud storage
- Databases
- APIs
- Web sources

Structured sources such as databases and APIs are often easier to process because their schemas and fields are known in advance.

Files are different. A document can contain text mixed with formatting, tables, images, multiple columns, scanned pages, and other information that the RAG system may need to interpret or discard.

That is why this module places particular emphasis on **document/file ingestion**.

## 4. Why Ingestion Is an Engineering Problem

Consider a PDF containing a policy document.

It may contain:

```text
Headings
Paragraphs
Tables
Images
Footnotes
Page numbers
Multiple columns
Scanned pages
```

The information we need for RAG may be mixed together with layout and presentation information.

A parser therefore has to determine which parts contain knowledge that should become searchable and how that information should be represented.

PDFs are particularly challenging because their internal representation is designed around rendering a page rather than providing a clean logical document hierarchy. Text can be represented through individual characters and coordinates, which means extraction may need to reconstruct words, lines, paragraphs, and columns.

## 5. Parsing Is Not Chunking

These two concepts are easy to confuse.

### Parsing

Parsing answers:

> **What useful information is actually inside this source document?**

For example:

```text
PDF
 ↓
Extract text
Extract tables
Extract metadata
Detect images/scanned pages
Preserve useful structure
```

### Chunking

Chunking answers:

> **How should the usable document content be divided into retrieval units?**

That is why we handle parsing first and chunking in the next module.

## 6. OCR: When the Document Is an Image

Not every document contains machine-readable text.

A scanned PDF may effectively be a collection of page images:

```text
Scanned Document
      ↓
   Page Image
      ↓
      OCR
      ↓
  Extracted Text
```

OCR becomes part of ingestion when ordinary text extraction cannot recover the information we need.

OCR quality can depend on factors such as image resolution, font complexity, and document layout. Errors introduced here can propagate into later stages of the RAG pipeline.

## 7. Cleaning Without Destroying Meaning

Extraction often produces noise.

For example, a parser might introduce unwanted line breaks, repeated headers, broken words, or other artifacts.

Cleaning should therefore be conservative.

The goal is **not**:

> Remove everything unusual.

The goal is:

> **Remove extraction noise while preserving information and structure that may matter later.**

This distinction becomes important when we work with headings, tables, lists, and other structured content.

## 8. Structure, Metadata, and Provenance

A useful ingestion pipeline should not leave us with an anonymous block of text.

We want the extracted content to remain traceable to its source.

A document representation might eventually carry information such as:

```text
document_id
source
page
section
title
document_type
created_at
updated_at
provenance
text
```

These fields can later support filtering, retrieval, citations, debugging, and document lifecycle management.

## 9. Validation: Did Ingestion Actually Work?

A pipeline finishing without an exception does **not** necessarily mean that ingestion succeeded.

We need to ask questions such as:

- Did we extract meaningful text?
- Is the output empty?
- Was the document partially extracted?
- Did OCR fail on important pages?
- Did table structure disappear?
- Can we trace the output back to the source?

Validation should therefore be a deliberate stage of ingestion, not an afterthought.

## 10. The Complete Mental Model

At the end of this module, the ingestion side of our RAG system should look like:

```text
                    DOCUMENT INGESTION

Raw Document
     ↓
Validate & Register
     ↓
Parse
     ↓
OCR if Required
     ↓
Clean & Normalize
     ↓
Preserve Structure
     ↓
Metadata + Provenance
     ↓
Validate
     ↓
Structured Document
     ↓
     └──────────────→ Module 04: Chunking
```

The important boundary is the final one: **Module 03 prepares reliable document content; Module 04 decides how that content should be divided for retrieval.**

## 11. Production Perspective

In a production RAG system, ingestion is not necessarily a one-time script.

Documents can be:

- Added
- Updated
- Deleted
- Replaced by newer versions

Production pipelines may therefore need asynchronous processing, retries, document identity, versioning, incremental updates, and visibility into processing state.

For very large collections, ingestion also becomes a systems problem involving throughput, document size, latency, cost, and update frequency.

We will introduce these concerns here, then return to them in the later production-oriented parts of the course.

## 12. A Simple Example

Suppose we receive:

```text
refund_policy.pdf
```

Our ingestion system should conceptually transform it into something like:

```text
Document
├── document_id: refund_policy
├── source: refund_policy.pdf
├── pages: ...
├── sections: ...
├── metadata: ...
└── extracted_content: ...
```

We are **not embedding it yet**.

We are **not storing vectors yet**.

We are **not retrieving it yet**.

We are making sure the source has been converted into a reliable representation that the next stage can work with.

## Key Takeaways

1. RAG has an **ingestion flow** and a **query flow**.
2. Document ingestion prepares external knowledge for downstream RAG stages.
3. Real-world files are harder to ingest than simple structured records because they can contain mixed formats, layout, images, tables, and scans.
4. **Parsing** extracts and structures useful information; **chunking** divides that information into retrieval units.
5. OCR is needed when important content is image-based or scanned.
6. Cleaning must remove noise without destroying meaning.
7. Metadata and provenance keep extracted knowledge traceable to its source.
8. Validation checks whether ingestion actually produced usable content.
9. The output of this module should be ready for **Module 04 — Chunking**.
